```python
'''Esercizio Memorizza e stampa tutti i fattori di un numero dato in input (cioè tutti i numeri primi che, moltiplicati tra di loro, danno il numero di partenza). 
Esempio: 
    • input: 150 
    • output: [2, 3, 5, 5]'''
num = int(input('Inserisci un numero: '))
fattore = []
divisore = 2

while num > 1:
    while num % divisore == 0:
        fattore.append(divisore)
        num //= divisore
    divisore += 1
print(fattore)
```

```python
'''Esercizio Calcolare (ma non stampare) le prime N potenze di K; ognuna di esse andrà memorizzata in coda a una lista. 
Alla fine, stampare la lista risultante. Proviamo con diversi valori di K, oppure facciamola inserire all'utente. 
Realizzare due versioni: 
    • con un ciclo while, 
    • con un ciclo for.'''
k = int(input("Inserisci il valore di K: "))
n = int(input("Inserisci il valore N: "))

#Ciclo while
nk_while = []
power = 0

while power < n:
    result = k ** power
    nk_while.append(result)
    power += 1

print("Lista generata:", nk_while)


#ciclo for
nk_for = []

for potenza in range(n):
    result = k ** potenza
    nk_for.append(result)

print("Lista generata:", nk_for)
```

```python
#Esercizio Abbiamo una lista di parole: 
parole = ["Albergo", "Sedia", "Borgo", "Petalo", "Eremo", "Belvedere", "Semestre", "Esteta", "Sosta", "Orpello", "Abete", "Orologio", "Cesta", "Ermellino"] 
#stampiamo, per ogni parola, quante volte appare la lettera "e"; facciamo attenzione al fatto che appare sia maiuscola che minuscola.

for parola in parole:
    conteggio = parola.lower().count("e")
    print(f"La parola '{parola}' contiene la lettera 'e' {conteggio} volte.")
```
