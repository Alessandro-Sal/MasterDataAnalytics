```python
#Esercizio L'azienda Object SpA ha creato una lista di quanti oggetti ha venduto ogni mese nell'ultimo anno: 
vendite = [2000, 5500, 7200, 4320, 1280, 1900, 2500, 3900, 6410, 8150, 7100, 5350] 
#trasformiamola in una Series di Pandas (casting):
import pandas as pd
vendite = pd.Series(vendite) 
print(vendite)
print("")
print("Analisi:")
print("")
#e rispondiamo alle domande del CEO della Object SpA: 
#• qual è stata la vendita massima mensile? E quella minima?
max_vendite = vendite.max()
print(f"La vendita massima è:",max_vendite)
min_vendite = vendite.min()
print("-")
print(f"La vendita minima è:",min_vendite)
#• quali sono le vendite mensili maggiori di 5000 oggetti? E quante ne sono?
print("-")
vendite_sopra_5000 = vendite[vendite>5000]
print(f"Le vendite sopra i 5000 sono:\n",vendite_sopra_5000)
print("-")
conteggio_sopra_5000 = vendite_sopra_5000.count()
print(f"Il numero delle vendite sopra i 5000 è:",conteggio_sopra_5000)
print("-")
#• quali sono le vendite minori di 3000 oggetti?
vendite_sotto_3000 = vendite[vendite<3000]
print(f"Le vendite sotto i 3000 sono:\n",vendite_sotto_3000)
print("-")
#• in media quanti oggetti sono stati venduti al mese?
vendite_conteggio = vendite.count()
vendite_somma = vendite.sum()
vendite_media = vendite_somma / vendite_conteggio
print(f"La media delle vendite è ",vendite_media)
```

```python
#Esercizio L'azienda Object SpA ha un dataset con tutti gli stipendi dei dipendenti, memorizzato in una Series: 
import pandas as pd 
stipendi = pd.Series( [100, 200, 300, 400, 500,  600, 700, 800, 900, 1000] ) 
#L'azienda ci chiede di raddoppiare tutti gli stipendi; facciamolo in due modi: • con un ciclo for 7 • con il masking
stipendi
#ciclo for
nuovi_stipendi_list = []
for stipendio in stipendi:
    nuovi_stipendi_list.append(stipendio * 2)
stipendi_raddoppiati_for = pd.Series(nuovi_stipendi_list)

print(stipendi_raddoppiati_for)
print("-")
#masking
maschera = stipendi > 0
stipendi_masked = stipendi.copy()
stipendi_masked[maschera] = stipendi_masked[maschera] * 2

print(stipendi_masked)
```

```python
#Esercizio Scrivere un programma che, data una lista di numeri, fornisca in output i tre numeri più grandi; 
#gestire il caso in cui la lista sia più corta di tre, e quando uno o più dei numeri selezionati sono uguali.
import random as r
numeri = []
def top_tre(numeri):
    top_tre = sorted(numeri, reverse=True)[:3]
    
    return top_tre
list_length = r.randint(1, 8)
random_list = [r.randint(1, 20) for _ in range(list_length)]

print(f"Abbiano generato questa lista di numeri random: {random_list}")

result = top_tre(random_list)
print(f"La top 3 dei numeri è: {result}")
```

```python
#Esercizio Scrivere un programma che: 
    #• in input acquisisce una lista di numeri e un numero K 
    #• in output, dovrà restituire la media di tutti i numeri nella lista maggiori o uguali a K 
    #• se non ce ne dovesse essere nessuno, dovrà stampare a schermo un messaggio adeguato.
def calcola_media_maggiori_uguali(lista_numeri, k):
    numeri_filtrati = [numero for numero in lista_numeri if numero >= k]
    if not numeri_filtrati:
        print("Non ci sono numeri filtrati")
        return None
    somma = sum(numeri_filtrati)
    quantita = len(numeri_filtrati)
    media = somma / quantita
    return media

numeri_di_prova = [4, 7, 12, 2, 9, 15]
valore_k = 8

risultato = calcola_media_maggiori_uguali(numeri_di_prova, valore_k)

if risultato is not None:
    print(f"Il calcolo medio è: {risultato}")
```

```python
#Esercizio Scrivere un programma che, data una lista di numeri, come output stamperà 
#lo stesso numero di asterischi su righe diverse, ottenendo una semplice visualizzazione grafica 
#Esempio, supponendo di avere il seguente input: 
numeri = [5, 2, 3, 4] 
#L'output sarà: ***** ** *** ****
def stampa_asterischi(lista_numeri):
    for numero in lista_numeri:
        riga_asterischi = "*" * numero
        print(riga_asterischi)
numeri_input = numeri

print("Output con asterischi:")

stampa_asterischi(numeri_input)
```

```python
#Esercizio Abbiamo una lista di codici fiscali: 
lista_cf = ["ABCDEF95G01A123B", "GHIJKL91M02A321C", "MNOPQR89S03A456D", "STUVWX95Z04A654E", "XYZABC01D05A789F", "DEFGHI95J06A987G"] 
#• trovare i codici fiscali che contengono "95", metterli in una lista, e alla fine stamparla; 
#• inoltre, per ognuno di essi, stampare a video i caratteri relativi al nome e quelli relativi al cognome.
def elabora_codici_fiscali(lista_cf):
    cf_con_95 = []
    print("Analizzando i CF")
    print("-" * 30)

    for cf in lista_cf:
        if "95" in cf:
            cf_con_95.append(cf)
            caratteri_cognome = cf[:3]
            caratteri_nome = cf[3:6]
            
            print(f"Corrispondenza trovata: {cf}")
            print(f" -> Caratteri Cognome: {caratteri_cognome}")
            print(f" -> Caratteri Nome: {caratteri_nome}")
            print("-" * 30)
            
    print(f"La lista finale dei codici contiene '95':\n{cf_con_95}")

elabora_codici_fiscali(lista_cf)
```

```python
#Esercizio Abbiamo una lista di stringhe di prezzi in dollari, che erroneamente sono stati scritti con il simbolo dell'euro: 
prezzi = ["100 €", "200 €", "500 €", "10 €", "50 €", "70 €"] 
#cambiare il simbolo dell'euro (€) in quello del dollaro ($) per ogni stringa nella lista; il risultato sarà memorizzato in un'altra lista.
def converti_simbolo_valuta(lista_prezzi):
    prezzi_dollari = []
    
    for prezzo in lista_prezzi:
        prezzo_aggiornato = prezzo.replace("€", "$")
        prezzi_dollari.append(prezzo_aggiornato)
        
    return prezzi_dollari
print(f"Prezzo originale: {prezzi}")
risultato = converti_simbolo_valuta(prezzi)

print(f"I prezzi modificati sono: {risultato}")
```

```python
#Esercizio Abbiamo una lista di studenti: 
studenti = ["Alex", "Bob", "Cindy", "Dan", "Emma", "Faith", "Grace", "Henry", "Isabelle", "John"] 
#vogliamo dividere gli studenti in due squadre per un campionato di Uno nel seguente modo: selezioneremo i nomi in posizione pari per un squadra,
#e i nomi in posizione dispari per l'altra. 
#Creiamo due liste per ogni squadra, e alla fine visualizziamole.
def dividi_squadre(lista_studenti):
    squadra_pari = lista_studenti[0::2]
    squadra_dispari = lista_studenti[1::2]
    return squadra_pari, squadra_dispari

squadra_1, squadra_2 = dividi_squadre(studenti)

print("Team creato:")
print(f"Team 1 (Pari): {squadra_1}")
print(f"Team 2 (Dispari): {squadra_2}")
```

```python
#Esercizio Abbiamo una lista con i guadagni degli ultimi 12 mesi (supponiamo da Gennaio a Dicembre): 
guadagni = [100, 90, 70, 40, 50, 80, 90, 120, 80, 20, 50, 50] 
#dobbiamo confrontare, stampando tutto a video, il guadagno di ogni mese con la media dei guadagni precedenti, 
#e specificare nell'output se il guadagno attuale è maggiore o minore della media dei precedenti. 
#Esempio di un possibile output: Mese 1: 100 € Mese 2: 90 € (media prec: 100 € - il guadagno attuale è minore)
def analizza_guadagni(lista_guadagni):
    for i, guadagno in enumerate(lista_guadagni):
        mese = i + 1 
        
        if i == 0:
            print(f"Mese {mese}: {guadagno} €")
        else:
            guadagni_precedenti = lista_guadagni[:i]

            media_precedente = sum(guadagni_precedenti) / len(guadagni_precedenti)
            
            if guadagno > media_precedente:
                confronto = "guadagno corrente è maggiore"
            elif guadagno < media_precedente:
                confronto = "guadagno corrente è minore"
            else:
                confronto = "guadagno corrente è uguale"
                
            print(f"Mese {mese}: {guadagno} € (prev avg: {media_precedente:.2f} € - {confronto})")

print("Analizzando i guadadni mensili...")
print("-" * 65)

analizza_guadagni(guadagni)
```
