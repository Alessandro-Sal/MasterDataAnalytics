```python
#Esercizio Dal database AdventureWorks estraiamo la tabella dimproduct 
    #• Sulla colonna DealerPrice, utilizzando il metodo .round(), arrotondiamo i valori alle due cifre decimali, e poi al valore intero più vicino 
    #• Utilizzando il metodo .clip(), facciamo in modo che i valori siano compresi tra un minimo di 0 e un massimo di 1000

import os
import pandas as pd
import pymysql
import dotenv

dotenv.load_dotenv(dotenv_path=".env", override=True)
username = os.getenv("username")
password = os.getenv("password")
hostname = os.getenv("hostname")
database = os.getenv("database")

connection = pymysql.connect(
    host=hostname,
    user=username,
    password=password,
    database=database,
    cursorclass=pymysql.cursors.DictCursor,
)

query = "SELECT * FROM dimproduct;"
cursor = connection.cursor()
cursor.execute(query)
result = cursor.fetchall()

cursor.close()
connection.close()

df = pd.DataFrame(result)
```

```python
df = df.drop_duplicates()
df['DealerPrice'].info
df['DealerPrice'] = df['DealerPrice'].fillna(0)
df['DealerPrice'].info
print('--- Sulla colonna DealerPrice, utilizzando il metodo .round(), arrotondiamo i valori alle due cifre decimali, e poi al valore intero più vicino ---')
print(df['DealerPrice'].round(2))
df['DealerPrice'] = df['DealerPrice'].round(0)
print("\n",df['DealerPrice'])

print('--- Utilizzando il metodo .clip(), facciamo in modo che i valori siano compresi tra un minimo di 0 e un massimo di 1000 ---')
df['DealerPrice'] = df['DealerPrice'].clip(lower=0, upper=1000)
print("\n",df['DealerPrice'])
print(df[['ProductKey', 'EnglishProductName', 'DealerPrice']].head(10))
print(df[['ProductKey', 'EnglishProductName', 'DealerPrice']].tail(10))
```

```python
#Esercizio Creiamo un DataFrame sintetico, che contiene i guadagni mensili di diverse annate, con il seguente codice:
import numpy as np
years = 5 
guadagni = pd.DataFrame({"Mese": list("GFMAMGLASOND"*years),
                         "Anno": np.repeat(list(range(years)), 12),
                         "Valore": np.random.randint(800, 5000, 12*years)}) 
    #• Calcola la somma cumulativa dei guadagni utilizzando il metodo .cumsum() 
    #• Come sopra, ma raggruppato per ogni anno usando prima un .groupby()

g = guadagni
print("La somma cumulativa totale sul valore è:\n",g["Valore"].cumsum())

print("La somma cumulativa per anno sul valore è:\n",g.groupby('Anno')['Valore'].cumsum())
```

```python
#Esercizio 1/2 Dal database AdventureWorks estraiamo la tabella dimcustomer 
    #• Trasformiamo i nomi dei clienti in modo che abbiano solo lettere minuscole, e i cognomi in modo che abbiano solo lettere maiuscole 
    #• Sulla colonna EmailAddress, utilizzando il metodo .str.split(), estraiamo nome utente e dominio 
    #• Sulla colonna Phone, estraiamo ogni parte del numero (ad es. da "1 (11) 500 555-0162" a ["1", "(11)", "500", "555-0162"]) 
    #• Utilizzando il metodo .str.contains(), estraiamo tutti gli indirizzi e-mail che contengono il numero "21"
connection = pymysql.connect(
    host=hostname,
    user=username,
    password=password,
    database=database,
    cursorclass=pymysql.cursors.DictCursor,
)

query_customer = "SELECT * FROM dimcustomer;"
cursor = connection.cursor()
cursor.execute(query_customer)
result_customer = cursor.fetchall()

cursor.close()
connection.close()

df_customer = pd.DataFrame(result_customer)
```

```python
#• Trasformiamo i nomi dei clienti in modo che abbiano solo lettere minuscole, e i cognomi in modo che abbiano solo lettere maiuscole 
df_customer['FirstName'] = df_customer['FirstName'].str.lower()
df_customer['LastName'] = df_customer['LastName'].str.upper()

#• Sulla colonna EmailAddress, utilizzando il metodo .str.split(), estraiamo nome utente e dominio 
df_customer[['Username', 'Domain']] = df_customer['EmailAddress'].str.split('@', expand=True)

#• Sulla colonna Phone, estraiamo ogni parte del numero (ad es. da "1 (11) 500 555-0162" a ["1", "(11)", "500", "555-0162"]) 
df_customer['Phone_Parts'] = df_customer['Phone'].str.split(' ')

#• Utilizzando il metodo .str.contains(), estraiamo tutti gli indirizzi e-mail che contengono il numero "21"
emails_21 = df_customer[df_customer['EmailAddress'].str.contains("21", na=False)]

# Verify the results
print(df_customer[['FirstName', 'LastName', 'EmailAddress', 'Username', 'Domain', 'Phone_Parts']].head())
```

```python
#Esercizio 1/2 Dai beginner_datasets carichiamo in un DataFrame il file facebook.csv, che contiene dei post con data di pubblicazione, tipo (foto, video, …) #e numero di reactions raccolte: 
    #• Con la funzione pd.to_datetime() convertiamo la colonna status_published in formato Timestamp 
    #• Utilizzando gli attributi .dt.year , .dt.month , .dt.day , .dt.dayofweek , .dt.dayofyear, ottieniamo informazioni specifiche sulle date delle transazioni, come l'anno, il mese, il giorno della settimana, il giorno dell'anno, eccetera
path_fb = "C:/Users/Alessandro Personale/Desktop/Dataset/Dataset/beginner_datasets/facebook.csv"
df_f = pd.read_csv(path_fb)
df_f

#• Con la funzione pd.to_datetime() convertiamo la colonna status_published in formato Timestamp
df_f['status_published'] = pd.to_datetime(df_f['status_published'])

#• Utilizzando gli attributi .dt.year , .dt.month , .dt.day , .dt.dayofweek , .dt.dayofyear, ottieniamo informazioni specifiche sulle date delle transazioni, come l'anno, il mese, il giorno della settimana, il giorno dell'anno, eccetera
df_f['Year'] = df_f['status_published'].dt.year
df_f['Month'] = df_f['status_published'].dt.month
df_f['Day'] = df_f['status_published'].dt.day
df_f['DayOfWeek'] = df_f['status_published'].dt.dayofweek
df_f['DayOfYear'] = df_f['status_published'].dt.dayofyear

risultato = ['status_published', 'Year', 'Month', 'Day', 'DayOfWeek', 'DayOfYear']
print(df_f[risultato].head())

#Esercizio 2/2 
    #• Estraiamo solo i post relativi al 2012 
    #• Estraiamo solo i post relativi a maggio 2018 
    #• Confrontiamo il numero di post pubblicati nei weekend rispetto al numero di post pubblicati nel resto della settimana 
    #• Troviamo il primo e ultimo post pubblicati in ogni anno 
    #• Quanti tipi di post ci sono? E quanti per ogni tipo?

#• Estraiamo solo i post relativi al 2012
posts_2012 = df_f[df_f['Year'] == 2012]

#• Estraiamo solo i post relativi a maggio 2018 
posts_may_2018 = df_f[(df_f['Year'] == 2018) & (df_f['Month'] == 5)]

#• Confrontiamo il numero di post pubblicati nei weekend rispetto al numero di post pubblicati nel resto della settimana 
df_f['Weekend'] = df_f['DayOfWeek'].isin([5, 6])
weekend_counts = df_f['Weekend'].value_counts()

#• Troviamo il primo e ultimo post pubblicati in ogni anno 
primo_ultimo_post = df_f.groupby('Year')['status_published'].agg(['min', 'max'])

#• Quanti tipi di post ci sono? E quanti per ogni tipo?
unique_types_count = df_f['status_type'].nunique()
posts_per_type = df_f['status_type'].value_counts()

print(f"--- Numeri di post nel 2012: {len(posts_2012)} ---")
print(f"--- Numeri di post a Maggio 2018: {len(posts_may_2018)} ---")

print("\n--- Weekdays vs Weekends ---")
print(weekend_counts)

print("\n--- Primo e ultimo Post per anno ---")
print(primo_ultimo_post)

print(f"\n--- il totale di post unici: {unique_types_count} ---")
print(posts_per_type)
```

```python
#Esercizio 1/2 Nel pacchetto os della standard library c'è la funzione os.listdir() che permette di avere la lista dei nomi di file all'interno di una directory; 
#senza input di default li cerca nella directory di lavoro corrente, altrimenti si può passare un path per esaminare una directory specifica, 
#ad esempio os.listdir("mio_progetto/beginner_datasets/") 
    #• Nella directory dei beginner_datasets, quali sono i dataset che contengono dati nulli?
    #• Dovremo usare un ciclo for per esaminare tutti i nomi dei file 2. 3. 4. 5. 6. 
    #• Dovremo selezionare solo i nomi di file con estensione .csv (quindi usare un costrutto if) 
    #• Nel corpo dovremo leggere di volta in volta il file in esame, e caricarlo in un DataFrame con la funzione .read_csv() 
    #• Sul DataFrame dovremo utilizzare il metodo .isna() per trovare la maschera booleana dei dati nulli 
    #• Dovremo contare i dati nulli, utilizzando .sum(); potremmo doverlo utilizzare più di una volta 
    #• Dovremo stampare, o memorizzare in una list, solo i nomi dei file che contengono dati nulli
path = "C:/Users/Alessandro Personale/Desktop/Dataset/Dataset/beginner_datasets/"

csv_list = os.listdir(path)

datasets_val_nulli = []

print("--- Analisi dei valori nulli nei dataset (beginner_dataset) ---\n")

for csv in csv_list:
    if csv.endswith('.csv'):
        full_path = os.path.join(path, csv)
        
        try:
            df_total = pd.read_csv(full_path)
            total_nulls = df_total.isna().sum().sum()
            
            if total_nulls > 0:
                datasets_val_nulli.append(csv)
                print(f"[X] {csv} contiene {total_nulls} valori nulli.")
            else:
                print(f"[V] {csv} non ha valori nulli.")       
                
        except Exception as e:
            print(f"[!] Non posso leggere {csv}: {e}")
        
print("\n--- Fine ---")
print("Datasets con valori nulli:", datasets_val_nulli)
```
