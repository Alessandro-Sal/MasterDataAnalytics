```python
#Esercizio Il dataset insurance.csv dei beginner_datasets contiene dati su caratteristiche e abitudini delle persone rispetto ai costi individuali per le cure mediche come premio per le assicurazioni sulla salute. 
    #• Visualizziamone le dimensioni, un'anteprima, e osserviamo i nomi di colonna; 
    #• Quali sono le medie di charges rispetto a region? Ci sono differenze significative?
    #• E rispetto a smoker? E a sex? 
    #• Quali sono i descrittori statistici di bmi? Quali sono minimo, media e massimo di charges rispetto ai diversi quartili dei valori di bmi? 
#Nota: è lo stesso esercizio della volta scorsa, ma stavolta da eseguire con il metodo .groupby()

df_insurance = pd.read_csv('insurance.csv')
#display(df_insurance)

print('--- Visualizziamone le dimensioni, un\'anteprima, e osserviamo i nomi di colonna; ---')
print("--- Dimensioni del dataset ---")
print(df_insurance.shape)
print("\n--- Nomi delle colonne ---")
print(list(df_insurance.columns))
print("\n--- Anteprima del dataset ---")
print(df_insurance.head(5))


print('\n--- Quali sono le medie di charges rispetto a region? Ci sono differenze significative? ---')
media_charges_region = df_insurance.groupby('region')['charges'].mean().round(2)
print(media_charges_region)

print('\n--- E rispetto a smoker? E a sex? ---')
media_charges_smoker = df_insurance.groupby('smoker')['charges'].mean().round(2)
media_charges_sex = df_insurance.groupby('sex')['charges'].mean().round(2)
print('Smoker :',media_charges_smoker)
print('\nSex:',media_charges_sex)

print('\n--- Quali sono i descrittori statistici di bmi? Quali sono minimo, media e massimo di charges rispetto ai diversi quartili dei valori di bmi?  ---')
print(df_insurance['bmi'].describe().round(2))
bmi_quartile = pd.qcut(df_insurance['bmi'], q=4, precision=2) #qcut divide i valori in 4 quartili
charges_by_bmi_quartile = df_insurance.groupby(bmi_quartile, observed=True)['charges'].agg(['min', 'mean', 'max']).round(2)
print(charges_by_bmi_quartile)
```

```python
#Esercizio Carichiamo in un DataFrame il dataset iris.csv dei beginner_datasets e calcoliamo: 
    #• La media della lunghezza dei petali di tutto il dataset 
    #• La media della lunghezza dei petali per ogni specie di Iris, utilizzando il metodo .groupby() 
    #• Media, minimo e massimo della larghezza dei sepali per ogni specie, utilizzando .groupby() e .agg()

df_iris = pd.read_csv('iris.csv')
#DataCleaning
#display(df_iris.info)
df_iris.drop_duplicates(inplace=True) #3 duplicati tolti
display(df_iris.info)

print('--- La media della lunghezza dei petali di tutto il dataset ---')
media_petali = df_iris['petal_length'].mean().round(2)
print(media_petali)

print('\n--- La media della lunghezza dei petali per ogni specie di Iris, utilizzando il metodo .groupby() ---')
media_petali_per_species = df_iris.groupby('species')['petal_length'].mean().round(2)
print(media_petali_per_species)

print('\n--- Media, minimo e massimo della larghezza dei sepali per ogni specie, utilizzando .groupby() e .agg() ---')
stats_sepali = df_iris.groupby('species')['sepal_width'].agg(['mean', 'min', 'max']).round(2)
print('La media della larghezza dei Sepali è:\n',stats_sepali)
```

```python
#Esercizio Nei beginner_datasets, il dataset wine.csv contiene delle analisi organolettiche su diverse qualità di vini: 
    #• Qual è la media di concentrazione alcolica per ogni qualità? Ci sono differenze? E rispetto alla media totale? 
    #• C'è differenza nella concentrazione alcolica per vini bianchi e vini rossi? 
    #• Rifacendo le analisi dei due punti precedenti ma per il pH, cambia qualcosa? 
    #• E per i solfati?

df_wine = pd.read_csv('wine.csv')
#DataCleaning
df_wine_cleaned = df_wine.drop_duplicates()
df_w = df_wine_cleaned

display(df_w)
print('--- Qual è la media di concentrazione alcolica per ogni qualità? Ci sono differenze? E rispetto alla media totale? ---')
media_concentrazione_qualita = df_w.groupby('quality')['alcohol'].mean().round(2)
media_totale = df_w['alcohol'].mean().round(2)
print("La media di concentrazione alcolica per ogni qualità è:\n",media_concentrazione_qualita,"\nTendenza generale è al rialzo: i vini di qualità superiore tendono ad avere concentrazioni alcoliche più elevate.")
print("La media di concentrazione alcolica totale è:\n",media_totale,"\nI vini di qualità inferiore (punteggi 3–5) scendono al di sotto di questa media generale, mentre i vini di qualità superiore (punteggi 6–9) si collocano al di sopra.")

print('--- C\'è differenza nella concentrazione alcolica per vini bianchi e vini rossi? ---')
media_concentrazione_vini = df_w.groupby('type')['alcohol'].mean().round(2)
display(media_concentrazione_vini)

print('--- Rifacendo le analisi dei due punti precedenti ma per il pH, cambia qualcosa? ---')
media_concentrazione_qualita_ph = df_w.groupby('quality')['pH'].mean().round(2)
media_totale_ph = df_w['pH'].mean().round(2)
media_concentrazione_vini_ph = df_w.groupby('type')['pH'].mean().round(2)
print("La media di concentrazione del ph per ogni qualità è:\n",media_concentrazione_qualita_ph,"Non ci sono particolari differenze")
print("La media di concentrazione alcolica totale è\n:",media_totale_ph,"Nessuna particolare differenza con il totale")
display(media_concentrazione_vini_ph,"Nessuna differenza tra vino bianco e rosso")

print('\n--- E per i solfati? ---')
media_concentrazione_qualita_solfati = df_w.groupby('quality')['sulphates'].mean().round(2)
media_totale_solfati = df_w['sulphates'].mean().round(2)
media_concentrazione_vini_solfati = df_w.groupby('type')['sulphates'].mean().round(2)
print("La media di concentrazione dei solfati per ogni qualità è:\n", media_concentrazione_qualita_solfati)
print("\nLa media di concentrazione dei solfati totale è:\n", media_totale_solfati)
display(media_concentrazione_vini_solfati)
print("C'è una differenza significativa: i vini rossi contengono livelli di solfati notevolmente più alti rispetto ai vini bianchi.")
```

```python
#Esercizio 1/3 Nei beginner_datasets, il file boston.csv contiene il Boston Housing Dataset, 
#che deriva dalle informazioni raccolte dal Census Service degli Stati Uniti sulle abitazioni nell'area di Boston. 
 #Di seguito vengono descritte le colonne del dataset: 
    #• CRIM - tasso di criminalità pro capite per città 
    #• ZN - proporzione di terreni residenziali suddivisi in zone per lotti superiori a 25.000 piedi quadrati. 
    #• INDUS - percentuale di acri di attività commerciali non al dettaglio per città. 
    #• CHAS - variabile dummy del fiume Charles (1 se il tratto costeggia il fiume; 0 altrimenti)
#Esercizio 2/3 
    #• NOX - concentrazione di ossidi di azoto (parti per 10 milioni). 
    #• RM - numero medio di stanze per abitazione 
    #• AGE - proporzione di unità abitative occupate da proprietari costruite prima del 1940 
    #• DIS - distanze ponderate da cinque centri occupazionali di Boston 
    #• RAD - indice di accessibilità alle autostrade radiali 
    #• TAX - aliquota dell'imposta fondiaria sul valore pieno per 10.000 dollari 
    #• PTRATIO - rapporto alunni-insegnanti per città 
    #• BLACK - la percentuale di neri per città 
    #• LSTAT - % di popolazione di condizione più bassa 11 
    #• MEDV - Valore mediano delle case, espresso in migliaia di dollari
#Esercizio 3/3 
    #• La media del prezzo delle case cambia a seconda della distanza dal fiume Charles? 
    #• Si nota una correlazione tra il tasso di criminalità e il valore delle abitazioni? Come si può spiegare il risultato? 
    #• Qual è la media del numero di stanze rispetto al rapporto alunni-insegnanti? E del valore delle case? Appare esserci una qualche correlazione? Come si può spiegare il risultato? 
    #• Rispetto all'accessibilità alle autostrade, cambia qualcosa la media delle età delle abitazioni? E del numero di stanze? E delle tasse?

df_b = pd.read_csv('boston.csv')
#Data cleaning
#eliminiamo i duplicati
df_b.drop_duplicates(inplace=True)
df_b.info()

print('--- La media del prezzo delle case cambia a seconda della distanza dal fiume Charles? ---')
media_medv_by_chas = df_b.groupby('chas')['medv'].mean()
print(media_medv_by_chas, "\nLe case vicino al fiume Charles hanno un valore medio nettamente superiore")

print('\n--- Si nota una correlazione tra il tasso di criminalità e il valore delle abitazioni? Come si può spiegare il risultato? ---')
corr_crim_medv = df_b['crim'].corr(df_b['medv'])
print("Crime-Value Correlation:", corr_crim_medv, "è una correlazione negativa: all'aumentare della criminalità, il valore delle abitazioni scende.")

print('\n--- Qual è la media del numero di stanze rispetto al rapporto alunni-insegnanti? E del valore delle case? Appare esserci una qualche correlazione? Come si può spiegare il risultato? ---')
corr_ptratio_rm = df_b['ptratio'].corr(df_b['rm'])
corr_ptratio_medv = df_b['ptratio'].corr(df_b['medv'])
print("PTRATIO-Rooms Correlation:", corr_ptratio_rm)
print("PTRATIO-Value Correlation:", corr_ptratio_medv)
print("Sono correlazioni negative: le zone con scuole più affollate (alto PTRATIO) presentano case più piccole e di minor valore.")

print('\n--- Rispetto all\'accessibilità alle autostrade, cambia qualcosa la media delle età delle abitazioni? E del numero di stanze? E delle tasse? ---')
avg_metrics_by_rad = df_b.groupby('rad')[['age', 'rm', 'tax']].mean()
print(avg_metrics_by_rad, "\nDifferenza netta per RAD=24: qui le case sono molto più vecchie e subiscono la tassazione massima.")
```

```python
#Esercizio 1/3 Abbiamo un DataFrame di dipendenti: 
employees_df = pd.DataFrame({ 'employee_id': [101, 102, 103, 104, 105], 
                              'name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma'], 
                              'department_id': [1, 2, 1, 2, 3] })

#Esercizio 2/3 E un DataFrame di dipartimenti: 
departments_df = pd.DataFrame({ 'department_id': [1, 2, 3], 
                                'department_name': ['HR', 'IT', 'Finance'], 
                                'location': ['New York', 'San Francisco', 'Chicago'] })

#Esercizio 3/3 Unire questi DataFrame in base alla colonna comune department_id, in modo da avere nel risultato informazioni sia sui dipendenti che sui dipartimenti, usando la funzione .merge() 
    #• Per ogni DataFrame, trasformare la colonna department_id nell'indice, facendo in modo che la modifica sia permanente; poi unire i due dataset mediante il metodo .join() 
    #• Ci sono differenze nel risultato? Quali? Perché?
display(employees_df)
display(departments_df)

merged_df = pd.merge(employees_df, departments_df, on='department_id')

print("Risultato con .merge():")
display(merged_df)

employees_df.set_index('department_id', inplace=True)
departments_df.set_index('department_id', inplace=True)

joined_df = employees_df.join(departments_df)

print("\nRisultato con .join():")
display(joined_df)

print("\nLe differenze sono visive e strutturali, \nNel DataFrame merged_df (creato con .merge()), la colonna department_id appare come una normale colonna del dataset e l\'indice della tabella è quello posizionale di default (0, 1, 2, 3, 4).")
print("\nNel DataFrame joined_df (creato con .join()), department_id non è più una colonna standard, ma è diventato l\'indice di riga (evidenziato visivamente spostato a sinistra o su un livello diverso quando stampato a schermo).")
```

```python
#Esercizio 
    #• Dal database AdventureWorksDW importiamo le tabelle dimemployee e dimemployeesalesterritory come DataFrame 
    #• Effettuiamo un join tra i due DataFrame usando le colonne EmployeeKey 
    #• Controlliamo la dimensione del DataFrame risultante: è quella attesa? 
    #• Importiamo ora la tabella dimsalesterritory ed effettuiamo un join tra questa e il DataFrame risultante della join precedente, usando le colonne SalesTerritoryKey • Su questo DataFrame contare quanti dipendenti ci sono per ogni paese (country) e per ogni regione (region) 
    #• Valutiamo la media del BaseRate per ogni paese: ci sono differenze?

dotenv.load_dotenv(dotenv_path=".env", override=True)
username = os.getenv("username")
password = os.getenv("password")
hostname = os.getenv("hostname")
database = os.getenv("database")

connection = pymysql.connect(
    host=hostname,
    user=username,
    password=password,
    database=database,
    cursorclass=pymysql.cursors.DictCursor,
)
cursor = connection.cursor()

query_employee = "SELECT * FROM dimemployee;"
query_territory = "SELECT * FROM dimemployeesalesterritory;"

cursor.execute(query_employee)
result_employee = cursor.fetchall()
df_employee = pd.DataFrame(result_employee)

cursor.execute(query_territory)
result_territory = cursor.fetchall()
df_territory = pd.DataFrame(result_territory)

display(df_employee)
display(df_territory)

mergeAdw = pd.merge(df_employee, df_territory, on='EmployeeKey')
display(mergeAdw.shape)

query_sales_territory = "SELECT * FROM dimsalesterritory;"
cursor.execute(query_sales_territory)
result_sales_territory = cursor.fetchall()
df_sales_territory = pd.DataFrame(result_sales_territory)

cursor.close()
connection.close()

final_mergeAdw = pd.merge(mergeAdw, df_sales_territory, on='SalesTerritoryKey')

employee_count = final_mergeAdw.groupby(['SalesTerritoryCountry', 'SalesTerritoryRegion'])['EmployeeKey'].count().reset_index(name='EmployeeCount')

display(employee_count)

avg_baserate = final_mergeAdw.groupby('SalesTerritoryCountry')['BaseRate'].mean().reset_index(name='AverageBaseRate')

display(avg_baserate)
```
