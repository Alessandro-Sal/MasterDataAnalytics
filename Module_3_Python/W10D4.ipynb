```python
#Esercizio Tra i beginner datasets scaricabili all'indirizzo 
#https://www.kaggle.com/datasets/ahmettezcantekin/beginner-datasets 
#selezioniamo amazon.csv, un dataset contenente una serie di recensioni su Amazon. 
    #• Valutiamo la dimensione del dataset 
    #• Osserviamo quali sono i nomi di colonna 
    #• Visualizziamo dieci righe a caso; 
    #• Il dataset è bilanciato, ovvero: il numero di recensioni positive è uguale a quello delle negative, oppure no?
import pandas as pd
df = pd.read_csv('amazon.csv')
print(df)
df.info()
print(df.head())
print("-")
print("---Valutiamo la dimensione del dataset---")
print(f"Le righe sono {df.shape[0]} mentre le colonne sono {df.shape[1]}")
print("-")
print("---Osserviamo quali sono i nomi di colonna---")
print(f"Il nome della prima colonna è {df.columns[0]} mentre il nome della seconda colonna è {df.columns[1]}.")
print("-")
print("---Visualizziamo dieci righe a caso---")
print(df.sample(n=10))
print("-")
print("---Il dataset è bilanciato, ovvero: il numero di recensioni positive è uguale a quello delle negative, oppure no?---")
print(df['Positive'].value_counts())
recensioni_positive = df[df['Positive'] == 1]
recensioni_negative = df[df['Positive'] == 0]
num_positive = len(recensioni_positive)
num_negative = len(recensioni_negative)
recensioni_totali = num_positive + num_negative
peso_positive = (num_positive / recensioni_totali) * 100
print(f"Le recensioni positive sono il {peso_positive}% del totale")
```

```python
#Esercizio Il dataset diabetes.csv raccoglie persone con diabete o meno, e il valore di diverse variabili fisiologiche dei pazienti. 
    #• Osserviamone le dimensioni e un'anteprima di cinque righe; 
    #• Prendiamoci un po' di tempo per dare un'occhiata ai metadati delle colonne; 
    #• Stampiamo dei descrittori statistici del dataset; 
    #• Selezioniamo i dati relativi a diverse fasce di età: 20-29, 30-39, 40-50; 
    #• Qual è la media della pressione sanguigna diastolica per le diverse fasce di età?
df_diabetes = pd.read_csv("diabetes.csv")
print("-")
print("---Osserviamone le dimensioni e un'anteprima di cinque righe;---")
print(df_diabetes.head())
print(df_diabetes.info())
print(df_diabetes.shape)
print("-")
print("---Prendiamoci un po' di tempo per dare un'occhiata ai metadati delle colonne;---")
print(df_diabetes.columns)
print(df_diabetes.dtypes)
print("-")
print("---Stampiamo dei descrittori statistici del dataset;---")
print(df_diabetes.describe())
print("-")
print("---Selezioniamo i dati relativi a diverse fasce di età: 20-29, 30-39, 40-50;---")
age_20_29 = df_diabetes[(df_diabetes['Age (years)'] >= 20) & (df_diabetes['Age (years)'] <= 29)]
age_30_39 = df_diabetes[(df_diabetes['Age (years)'] >= 30) & (df_diabetes['Age (years)'] <= 39)]
age_40_50 = df_diabetes[(df_diabetes['Age (years)'] >= 40) & (df_diabetes['Age (years)'] <= 50)]
print(f"righe per fascia 20-29: {len(age_20_29)}")
print(f"righe per fascia 30-39: {len(age_30_39)}")
print(f"righe per fascia for 40-50: {len(age_40_50)}")
print("-")
print("---Qual è la media della pressione sanguigna diastolica per le diverse fasce di età?---")
media_bp_20_29 = age_20_29['Diastolic blood pressure (mm Hg)'].mean()
media_bp_30_39 = age_30_39['Diastolic blood pressure (mm Hg)'].mean()
media_bp_40_50 = age_40_50['Diastolic blood pressure (mm Hg)'].mean()

print(f"Pressione diastolica BP (20-29): {media_bp_20_29:.2f}")
print(f"Pressione diastolica BP (30-39): {media_bp_30_39:.2f}")
print(f"Pressione diastolica BP (40-50): {media_bp_40_50:.2f}")
```

```python
#Esercizio Il dataset insurance.csv contiene dati rispetto a caratteristiche e abitudini delle persone, e della zona in cui vivono, rispetto ai costi individuali per le cure mediche come premio per le assicurazioni sulla salute. 
    #• Visualizziamone le dimensioni, un'anteprima, e osserviamo i nomi di colonna; 
    #• Quali sono le medie di charges rispetto a region? Ci sono differenze significative? 
    #• E rispetto a smoker? E a sex? 
    #• Quali sono i descrittori statistici di bmi? Quali sono minimo, media e massimo di charges rispetto ai diversi quartili dei valori di bmi?
df_insurance = pd.read_csv('insurance.csv')
df_insurance
print("-")
print("---Visualizziamone le dimensioni, un'anteprima, e osserviamo i nomi di colonna;---")
print("Dimensioni:", df_insurance.shape)
print("Colonne:", df_insurance.columns.tolist())
print("Anteprima:")
print(df_insurance.head())
print("-")
print("---Quali sono le medie di charges rispetto a region? Ci sono differenze significative? ---")
print(df_insurance.groupby('region')['charges'].mean())
print("-")
print("---E rispetto a smoker? E a sex?---")
print(df_insurance.groupby('smoker')['charges'].mean())
print(df_insurance.groupby('sex')['charges'].mean())
print("-")
print("---Quali sono i descrittori statistici di bmi? Quali sono minimo, media e massimo di charges rispetto ai diversi quartili dei valori di bmi?---")
print(df_insurance['bmi'].describe())
df_insurance['bmi_quartile'] = pd.qcut(df_insurance['bmi'], q=4)
print(df_insurance.groupby('bmi_quartile', observed=True)['charges'].agg(['min', 'mean', 'max']))
print("-")
```

```python
#Esercizio 1/2 Il dataset pokemon.csv contiene un database di Pokémon, con dati quali nome, tipi di attacco, valori di attacco/difesa/eccetera, e se sono o meno leggendari. 
    #• Verifichiamo la dimensione, un'anteprima e osserviamo i nomi di colonna; 
    #• È verosimile che la prima colonna dovrebbe essere un indice? 
    #• Confrontiamolo con l'indice messo automaticamente da Pandas: combaciano? 
    #• Se no, settare la prima colonna come indice.
    #• Quali sono i Pokémon leggendari? 
    #• E quali sono i leggendari di tipo 1 Grass? 
    #• E leggendari di tipo 1 Ice o Fire? 
    #• Trasformiamo la colonna Name nell'indice; 
    #• Quali sono i Pokémon della prima generazione con attacco > 50 e HP < 60?
df_pokemon = pd.read_csv('pokemon.csv')
df_pokemon
print("-")
print("---Verifichiamo la dimensione, un'anteprima e osserviamo i nomi di colonna;---")
print("Dimensioni:", df_pokemon.shape)
print("Colonne:", df_pokemon.columns.tolist())
print("Anteprima:")
print(df_pokemon.head())
print("-")
print("---È verosimile che la prima colonna dovrebbe essere un indice? Confrontiamolo con l'indice messo automaticamente da Pandas: combaciano? ---")
print("Pandas Index:", df_pokemon.index)
print("First column:", df_pokemon.columns[0])
print("-")
print("---Se no, settare la prima colonna come indice.---")
df_pokemon.set_index(df_pokemon.columns[0], inplace=True)
print("\n--- Dataset dopo aver settato la prima colonna come indice ---")
print(df_pokemon.head())
print("-")
print("---Quali sono i Pokémon leggendari?---")
Pokemon_legendary = df_pokemon[df_pokemon['Legendary'] == True]
print(Pokemon_legendary.head())
print("-")
print("---E quali sono i leggendari di tipo 1 Grass?---")
Grass_legendary = Pokemon_legendary[Pokemon_legendary['Type 1'] == 'Grass']
print(Grass_legendary.head())
print("-")
print("---E leggendari di tipo 1 Ice o Fire?---")
Ice_Fire_legendary = Pokemon_legendary[(Pokemon_legendary['Type 1'] == 'Ice') | (Pokemon_legendary['Type 1'] == 'Fire')]
print(Ice_Fire_legendary.head())
print("-")
print("---Trasformiamo la colonna Name nell'indice---")
df_pokemon.set_index('Name', inplace=True)
print("\n--- Dataset dopo aver settato il nome del pokemon come indice ---")
print(df_pokemon.head())
print("-")
print("---Quali sono i Pokémon della prima generazione con attacco > 50 e HP < 60?---")
gen1_att_hp = df_pokemon[(df_pokemon['Generation'] == 1) & (df_pokemon['Attack'] > 50) & (df_pokemon['HP'] < 60)]
print("\n--- Gen 1, Attack > 50, HP < 60 ---")
print(gen1_att_hp)
```

```python
#Esercizio 1/2 Dai dataset in piattaforma, carichiamo il file con i dati dei pub, circoli e locali in Italia, chiamato Mappa-dei-pub-circoli-locali-in-Italia.csv 
#Nota: il dataset non ha un encoding UTF-8 ma Latin1; inoltre il separatore non è una virgola, ma un punto e virgola. 
#Dunque per leggerlo dovremo aggiungere i parametri encoding= e sep=, ad esempio: 
#pd.read_csv(file_path, encoding="latin1", sep=";") dove file_path è la posizione del file nel nostro calcolatore.
#Esercizio 2/2 Esaminiamo il dataset: 
    #• quanti dati ci sono in totale? 
    #• quali sono i metadati? 
    #• stampiamo il primo elemento 
    #• stampiamo l'ultimo elemento 
    #• riusciamo a stampare un elemento a caso? 
    #• quali sono gli anni di inserimento presenti? 
    #• quante enoteche ci sono, e come si chiamano? L’ultimo punto potrebbe richiedere molti passaggi, perché i dati in questo dataset non hanno subìto un processo di pulizia e trasformazione adeguato!
df_pub = pd.read_csv('Mappa-dei-pub-circoli-locali-in-Italia.csv', encoding="latin1", sep=";")
print(df_pub)
print('------')
print('---Quanti dati ci sono in totale?---')
print(f"Ci sono {len(df_pub)} records, di cui {df_pub.shape[0]} righe e {df_pub.shape[1]} colonne")
print('------')
print('---Quali sono i metadati?---')
df_pub.info()
print('------')
print('---stampiamo il primo elemento---')
print(df_pub.iloc[0])
print('------')
print('---stampiamo l\'ultimo elemento---')
print(df_pub.iloc[-1])
print('------')
print('---riusciamo a stampare un elemento a caso?---')
print(df_pub.sample(1))
print('------')
print('---quali sono gli anni di inserimento presenti?---')
anni_unici = df_pub['Anno inserimento'].unique()
for anno_unico in anni_unici:
    print('Anno unico',anno_unico)
print('------')
print('---quante enoteche ci sono, e come si chiamano? L’ultimo punto potrebbe richiedere molti passaggi, perché i dati in questo dataset non hanno subìto un processo di pulizia e trasformazione adeguato!---')    
filtro_enoteche = df_pub['Nome'].str.contains('enoteca', case=False, na=False)
df_enoteche = df_pub[filtro_enoteche]
print(f"Le enoteche sono:{len(df_enoteche)} e il nome delle enoteche è:{df_enoteche['Nome'].tolist()}")
```
